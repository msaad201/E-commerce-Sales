# Data Understanding

| Column | What it means |
|---|---|
| `title` | The product name and description as shown on Amazon. |
| `rating` | The average customer star rating, out of 5. |
| `number_of_reviews` | How many customers reviewed the product. |
| `bought_in_last_month` | Approximate number of units bought in the past month, such as "100+" or "2K+". |
| `current/discounted_price` | The price the customer pays right now. |
| `price_on_variant` | The price of the default variant (size, pack, color, etc.) of the product. |
| `listed_price` | The original price before any discount. It says "No Discount" if there isn't one. |
| `is_best_seller` | Any badge on the product, like "Best Seller", "Amazon's Choice" or "Limited time deal". "No Badge" means none. |
| `is_sponsored` | Whether it is a paid ad ("Sponsored") or a normal search result ("Organic"). |
| `is_couponed` | Whether a coupon is available and how much it saves, like "Save 15% with coupon". |
| `buy_box_availability` | Whether the product has an "Add to cart" button. Empty means it can't be bought directly. |
| `delivery_details` | The estimated delivery date shown for the product. |
| `sustainability_badges` | Extra labels such as "Small Business", "Carbon impact" or "Works with Alexa". Mostly empty. |
| `image_url` | The link to the product's picture. |
| `product_url` | The link to the product's page on Amazon. |
| `collected_at` | The date and time the data was scraped. |

# Feature Engineering

#### Here I am going to create saperated columns to get delivery time, paid or free From the column 'delivery_details' and Products type from the column 'product_url'.

```python
print(df["delivery_details"].value_counts())
```

```python
# Here i got the delivery date year,month and day and created a new column called 'delivery_date'.  
delivery_date = df["delivery_details"].str.extract(r"((?:Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)\s+\d{1,2})",expand = False)
delivery_date
```

```python
df['Order_Placed'] = pd.to_datetime(df['Order_Placed'])
delivery_year = df['Order_Placed'].dt.year.astype(str)

df['delivery_days'] = pd.to_datetime(delivery_date + ' '+delivery_year)
```

```python
df['delivery_days']
```

```python
df['delivery_days'] = (df['delivery_days'] - df['Order_Placed'].dt.normalize()).dt.days
```

```python
df['delivery_days']= df['delivery_days'].fillna(0)
```

```python
px.histogram(data_frame=df , x = df['delivery_days']).show()
```

```python
def shipping(x):
    if pd.isna(x):
        return 'Unknown'
    elif "FREE" in x :
        return 'Free'
    else:
        return 'Paid'

df['delivery_details']=df['delivery_details'].apply(shipping)
```

```python
df= df.rename(columns= {'delivery_details':'delivery'})
```
